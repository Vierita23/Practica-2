## Regresion lineal 

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    "horas" : [1,2,3,4,5],
    "notas" : [2,3,5,4,6]
})

X= df[["horas"]]
Y= df[["notas"]]
print(X)
```

```python
modelo = LinearRegression()
modelo.fit(X, Y)
```

```python
print("Interceptor: ", modelo.intercept_)
print("Pendiente: ", modelo.coef_[0])
```

```python
y_pred = modelo.predict(X)
```

```python
horas = float(input("Ingrese las horas de estudio"))
prediccion = modelo.predict([[horas]])
print("Prediccion: ", prediccion)
```

```python
# Generamos la grafica 

plt.scatter(X, Y, color="blue", label="Datos reales")
plt.plot(X, y_pred, color="red", label="Regresion lineal")

plt.xlabel("horas de estudio")
plt.ylabel("nota")
plt.title("Notas de estudio")

plt.legend()
plt.grid(True)
plt.show()
```
